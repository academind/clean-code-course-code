```python
class Coordinate:
    def __init__(self, x, y):
        self.x = x
        self.y = y


class Rectangle:
    def __init__(self, origin, width, height):
        self.origin = origin
        self.width = width
        self.height = height

    def calculate_area(self):
        return self.width * self.height

    def end_point_top_right_x(self):
        return self.origin.x + self.width

    def end_point_bottom_left_y(self):
        return self.origin.y + self.height

    def print_end_points(self):
        print("End Point X-Axis (Top Right): " + str(self.end_point_top_right_x()))
        print("End Point Y-Axis (Bottom Left): " + str(self.end_point_bottom_left_y()))


def create_rectangle():
    main_position = Coordinate(50, 100)
    rectangle_position = Rectangle(main_position, 90, 10)

    return rectangle_position


rectangle = create_rectangle()

print(rectangle.calculate_area())
rectangle.print_end_points()
```
